###문제 : 다음 주어진 데이터 세트 분석을 통해
####1)가장 많은 학생이 수강한 과목을 찾아 결측치를 제거한 후
####2)해당과목의 점수를 표준화하여 가장 높은 점수를 소수 둘째 자리로 반올림하여 나타내 보시오

####1.데이터 불러오기

```python
import pandas as pd # 데이터 로드 및 전처리에 사용

df = pd.read_csv("DSMH_2/_10_8/students_scores.csv")

print(df.head(3), '\n\n')
print(df.tail(3))
```

####2.데이터 정보 알아보기

```python
#데이터 정보 알아보기
print(df.info())
```

####3.수치형 데이터 기술통계 확인하기

```python
print(df.describe())
```

####4.과목별 수강 카운트 하기

```python
#과목별 수강 카운트 하기
# student_id 칼럼을 제외한 df1 생성
df1 = df[['Statistics','DataScience','MachineLearning','EnglishLiterature']]

subject_counts = df1.count()
print(subject_counts)
```

####5.가장 많이 들은 과목이름 찾기

```python
#가장 많이 들은 과목이름
most_popular_subject= subject_counts.idxmax()
print(most_popular_subject)
```

####6.가장많이 들은 과목의 결측치가 있을 때 제거하기

```python
mps_df = df1[most_popular_subject].dropna()
print(mps_df.info())
```

####7.비교대상이 여러과목이므로 과목점수를 표준화 해 준다

```python
#scipy.stats 함수 이용 시 결과 같음
import scipy.stats as stats
mps_score = stats.zscore(mps_df)
print(mps_score[:10])
```

```python
#sklearn 라이브러리 함수이용 시 결과 같음
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(mps_df.values.reshape(-1,1))
mps_score = scaler.transform(mps_df.values.reshape(-1,1))
print(mps_score[:10])
```

```python
#계산하기 식: (x-mean(x))/std())

mps_score = (mps_df - mps_df.mean()) / mps_df.std()
print(mps_score[:10])
```

####8.표준화 점수중 최대값 찾기

```python
result = round(max(mps_score),2)
print(result)
```
