### 5. sklearn 제공 iris 데이터셋 불러오기

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

iris = load_iris()
iris_data = iris.data
iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)

print(iris_df.info())
print('\nfeature 들의 평균 값 : \n\n', iris_df.mean())
print('\nfeature 들의 분산 값 : \n\n', iris_df.var())

```

### 6.  iris 데이터셋 스케일링: 표준화 해보기

```python
scaler = StandardScaler()
scaler.fit(iris_df)
iris_scaled = scaler.transform(iris_df)

iris_df_scaled = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)


print('\nfeature 들의 평균 값 : \n\n', iris_df_scaled.mean())
print('\nfeature 들의 분산 값 : \n\n', iris_df_scaled.var())
```

### 7.  iris 데이터셋 스케일링: 정규화 해보기

```python
from sklearn.preprocessing import MinMaxScaler
# import pandas as pd   :  필수항목  5번에서 불러왔으므로 생략
# 최소 -최대 정규화 방법을 사용하면 모든 값은 최소값 0과 최대값 1사이에 위치한다

scaler_MM = MinMaxScaler()
scaler_MM.fit(iris_df)
iris_scaled_MM = scaler_MM.transform(iris_df)

iris_df_scaled_MM = pd.DataFrame(data=iris_scaled_MM, columns=iris.feature_names)

print('\nfeature 들의 최소 값 : \n\n', iris_df_scaled_MM.min())
print('\nfeature 들의 최대 값 : \n\n', iris_df_scaled_MM.max())

```
